아래 텍스트·속도·파일명을 바꿔 실행합니다. 첫 실행은 한국어 모델과 토크나이저를 내려받으므로 시간이 걸립니다.

```python
text = "태초에, 신은 세상을 창조했다."
speed = 0.95
output_path = "/content/intro_test.wav"
```

```python
from pathlib import Path
import json, subprocess

job = Path('/content/melotts_job.py')
job.write_text('''from melo.api import TTS
import json
from pathlib import Path
config = json.loads(Path("/content/melotts_job.json").read_text(encoding="utf-8"))
model = TTS(language="KR", device="cpu")
speakers = model.hps.data.spk2id
model.tts_to_file(config["text"], speakers["KR"], config["output"], speed=config["speed"])
''', encoding='utf-8')
Path('/content/melotts_job.json').write_text(json.dumps({'text': text, 'speed': speed, 'output': output_path}, ensure_ascii=False), encoding='utf-8')
subprocess.run(['/content/melotts-venv/bin/python', str(job)], check=True)
print(output_path)
```
